```python
from google.colab import files

uploaded = files.upload()

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

print(df.head())
print(df.shape)
print(df.dtypes)
```

```python
print(df.describe())
```

```python
high_adoption = df[
    df["Proportion of Energy from Renewables"] > 50
]

print(high_adoption.head())
```

```python
india = df[
    df["Country"] == "India"
]

print(india.head())
```

```python
sorted_df = df.sort_values(
    by="Proportion of Energy from Renewables",
    ascending=False
)

print(sorted_df.head())
```

```python
adoption = df[
    "Proportion of Energy from Renewables"
]

print("Mean :", adoption.mean())

print("Median :", adoption.median())

print("Mode :")
print(adoption.mode())

print("Standard Deviation :", adoption.std())
```

```python
country_avg = (
    df.groupby("Country")
    ["Proportion of Energy from Renewables"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(country_avg)
```

```python
country_avg.sort_values(
    ascending=False
).plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "Top 10 Countries by Average Renewable Energy Adoption"
)

plt.ylabel(
    "Average Renewable Energy Share (%)"
)

plt.xlabel("Country")

plt.xticks(rotation=45)

plt.show()
```

```python
corr = df[
    [
        "Production (GWh)",
        "Installed Capacity (MW)",
        "Investments (USD)",
        "Population",
        "GDP",
        "Energy Consumption",
        "CO2 Emissions",
        "Renewable Energy Jobs",
        "R&D Expenditure",
        "Energy Storage Capacity",
        "Electricity Prices",
        "Energy Subsidies",
        "Proportion of Energy from Renewables"
    ]
].corr()

print(corr)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title(
    "Correlation Matrix of Global Renewable Energy Factors"
)

plt.show()
```

```python
yearly_adoption = (
    df.groupby("Year")
    ["Proportion of Energy from Renewables"]
    .mean()
)

plt.figure(figsize=(10, 6))

plt.plot(
    yearly_adoption.index,
    yearly_adoption.values,
    marker="o"
)

plt.title(
    "Global Renewable Energy Adoption Trend"
)

plt.xlabel("Year")

plt.ylabel(
    "Average Renewable Energy Share (%)"
)

plt.grid(True)

plt.show()
```

```python
yearly_production = (
    df.groupby("Year")
    ["Production (GWh)"]
    .sum()
)

plt.figure(figsize=(10, 6))

plt.plot(
    yearly_production.index,
    yearly_production.values,
    marker="o"
)

plt.title(
    "Global Renewable Energy Production Trend"
)

plt.xlabel("Year")

plt.ylabel(
    "Renewable Energy Production (GWh)"
)

plt.grid(True)

plt.show()
```
