```python
# ============================================================
# CRISP-DM - GLOBAL RENEWABLE ENERGY ADOPTION ANALYSIS
# ============================================================

# ------------------------------------------------------------
# Import Libraries
# ------------------------------------------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ------------------------------------------------------------
# PHASE 1: BUSINESS UNDERSTANDING
# ------------------------------------------------------------
print("=" * 70)
print("PHASE 1: BUSINESS UNDERSTANDING")
print("=" * 70)

print("""
Objective:
Analyze global renewable energy adoption and predict whether
a country has HIGH or LOW renewable energy adoption.

Business Goals:
1. Understand global renewable energy trends.
2. Identify countries with high renewable adoption.
3. Analyze factors influencing renewable adoption.
4. Study relationships between renewable energy and economic,
   environmental and infrastructure indicators.
5. Build a Random Forest model to predict high renewable adoption.
""")

# ------------------------------------------------------------
# Upload Dataset
# ------------------------------------------------------------
print("\nUpload your Kaggle CSV file:")
uploaded = files.upload()

# Automatically detect CSV file
csv_files = [f for f in uploaded.keys() if f.lower().endswith(".csv")]

if len(csv_files) == 0:
    raise FileNotFoundError("No CSV file found. Please upload the Kaggle CSV dataset.")

file_name = csv_files[0]

print("\nUsing file:", file_name)

# ------------------------------------------------------------
# PHASE 2: DATA UNDERSTANDING
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("PHASE 2: DATA UNDERSTANDING")
print("=" * 70)

df = pd.read_csv(file_name)

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe(include="all").T)

print("\nColumn Names:")
for i, col in enumerate(df.columns):
    print(i, ":", col)

# ------------------------------------------------------------
# Detect Important Columns
# ------------------------------------------------------------
def find_column(columns, keywords):
    for col in columns:
        col_lower = col.lower().replace("_", " ").replace("-", " ")
        for keyword in keywords:
            if keyword in col_lower:
                return col
    return None


country_col = find_column(
    df.columns,
    ["country", "countries"]
)

year_col = find_column(
    df.columns,
    ["year"]
)

renewable_col = find_column(
    df.columns,
    [
        "proportion of energy from renewables",
        "renewable energy proportion",
        "renewable proportion",
        "renewable share",
        "renewable energy share",
        "renewables share",
        "renewable adoption"
    ]
)

print("\nDetected Columns:")
print("Country column  :", country_col)
print("Year column     :", year_col)
print("Renewable column:", renewable_col)

if renewable_col is None:
    raise ValueError(
        "Could not automatically identify the renewable-energy proportion column.\n"
        "Check the printed column names and replace renewable_col with the exact column name."
    )

# ------------------------------------------------------------
# DATA VISUALIZATION - UNDERSTANDING
# ------------------------------------------------------------

# 1. Missing Values
plt.figure(figsize=(14, 6))
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

if len(missing) > 0:
    missing.plot(kind="bar")
    plt.title("Missing Values by Column")
    plt.xlabel("Columns")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=90)
    plt.tight_layout()
    plt.show()
else:
    print("\nNo missing values found.")

# 2. Renewable Adoption Distribution
plt.figure(figsize=(10, 6))
sns.histplot(
    df[renewable_col].dropna(),
    bins=30,
    kde=True
)
plt.title("Distribution of Renewable Energy Adoption")
plt.xlabel(renewable_col)
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# PHASE 3: DATA PREPARATION
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("PHASE 3: DATA PREPARATION")
print("=" * 70)

# Make a copy
data = df.copy()

# ------------------------------------------------------------
# Convert Year
# ------------------------------------------------------------
if year_col is not None:
    data[year_col] = pd.to_numeric(
        data[year_col],
        errors="coerce"
    )

# ------------------------------------------------------------
# Convert Renewable Target Column to Numeric
# ------------------------------------------------------------
data[renewable_col] = pd.to_numeric(
    data[renewable_col],
    errors="coerce"
)

# ------------------------------------------------------------
# Remove rows where target is missing
# ------------------------------------------------------------
data = data.dropna(subset=[renewable_col]).copy()

# ------------------------------------------------------------
# Create High Renewable Adoption Target
# ------------------------------------------------------------
median_renewable = data[renewable_col].median()

data["High_Renewable_Adoption"] = (
    data[renewable_col] > median_renewable
).astype(int)

print("\nGlobal Median Renewable Adoption:")
print(round(median_renewable, 4))

print("\nTarget Distribution:")
print(data["High_Renewable_Adoption"].value_counts())

# ------------------------------------------------------------
# Remove original target column
# ------------------------------------------------------------
data.drop(
    columns=[renewable_col],
    inplace=True
)

# ------------------------------------------------------------
# Remove unnecessary ID-like columns
# ------------------------------------------------------------
id_columns = []

for col in data.columns:
    unique_ratio = data[col].nunique(dropna=False) / len(data)

    if (
        "id" in col.lower()
        and unique_ratio > 0.5
    ):
        id_columns.append(col)

if id_columns:
    print("\nRemoving ID-like columns:", id_columns)
    data.drop(columns=id_columns, inplace=True)

# ------------------------------------------------------------
# Handle Date columns if present
# ------------------------------------------------------------
for col in list(data.columns):

    if "date" in col.lower():

        try:
            date_values = pd.to_datetime(
                data[col],
                errors="coerce"
            )

            data[col + "_Year"] = date_values.dt.year
            data[col + "_Month"] = date_values.dt.month

            data.drop(columns=[col], inplace=True)

        except:
            pass

# ------------------------------------------------------------
# Handle Missing Values
# ------------------------------------------------------------
for col in data.columns:

    if col == "High_Renewable_Adoption":
        continue

    if data[col].dtype == "object":

        mode_value = data[col].mode()

        if len(mode_value) > 0:
            data[col] = data[col].fillna(mode_value[0])
        else:
            data[col] = data[col].fillna("Unknown")

    else:

        data[col] = data[col].fillna(
            data[col].median()
        )

# ------------------------------------------------------------
# Encode Categorical Columns
# ------------------------------------------------------------
label_encoders = {}

categorical_columns = data.select_dtypes(
    include="object"
).columns

for col in categorical_columns:

    encoder = LabelEncoder()

    data[col] = encoder.fit_transform(
        data[col].astype(str)
    )

    label_encoders[col] = encoder

print("\nCategorical columns encoded:")
print(list(categorical_columns))

# ------------------------------------------------------------
# Separate Features and Target
# ------------------------------------------------------------
X = data.drop(
    columns=["High_Renewable_Adoption"]
)

y = data["High_Renewable_Adoption"]

# ------------------------------------------------------------
# Remove constant columns
# ------------------------------------------------------------
constant_columns = [
    col for col in X.columns
    if X[col].nunique() <= 1
]

if constant_columns:
    print("\nRemoving constant columns:", constant_columns)
    X.drop(columns=constant_columns, inplace=True)

# ------------------------------------------------------------
# Convert to numeric
# ------------------------------------------------------------
X = X.apply(
    pd.to_numeric,
    errors="coerce"
)

# ------------------------------------------------------------
# Imputation
# ------------------------------------------------------------
imputer = SimpleImputer(
    strategy="median"
)

X = imputer.fit_transform(X)

# ------------------------------------------------------------
# Feature Scaling
# ------------------------------------------------------------
scaler = StandardScaler()

X = scaler.fit_transform(X)

# ------------------------------------------------------------
# Train-Test Split
# ------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])

# ------------------------------------------------------------
# PHASE 4: MODELING
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("PHASE 4: MODELING")
print("=" * 70)

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

model.fit(
    X_train,
    y_train
)

print("\nRandom Forest model trained successfully.")

# ------------------------------------------------------------
# PHASE 5: EVALUATION
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("PHASE 5: EVALUATION")
print("=" * 70)

y_pred = model.predict(X_test)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(
    "\nRandom Forest Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "Low Adoption",
            "High Adoption"
        ]
    )
)

# ------------------------------------------------------------
# Confusion Matrix
# ------------------------------------------------------------
cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Low Adoption",
        "High Adoption"
    ],
    yticklabels=[
        "Low Adoption",
        "High Adoption"
    ]
)

plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Feature Importance
# ------------------------------------------------------------
feature_names = data.drop(
    columns=["High_Renewable_Adoption"]
).columns

# Remove constant columns from feature names
feature_names = [
    col for col in feature_names
    if col not in constant_columns
]

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": model.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

print("\nTop 15 Important Features:")
display(
    importance_df.head(15)
)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=importance_df.head(15),
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 15 Features Influencing Renewable Energy Adoption"
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# PHASE 6: DEPLOYMENT
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("PHASE 6: DEPLOYMENT")
print("=" * 70)

print("""
The trained Random Forest model can be saved and deployed
for predicting whether a new country/year observation has
High or Low renewable energy adoption.
""")

# Optional model saving
import joblib

joblib.dump(
    model,
    "Global_Renewable_Energy_RandomForest.pkl"
)

joblib.dump(
    scaler,
    "Renewable_Energy_Scaler.pkl"
)

joblib.dump(
    imputer,
    "Renewable_Energy_Imputer.pkl"
)

print("\nModel saved successfully.")

# ------------------------------------------------------------
# ADDITIONAL VISUALIZATIONS
# ------------------------------------------------------------

# Country-wise average renewable adoption
if country_col is not None:

    country_analysis = (
        df.groupby(country_col)[renewable_col]
        .mean()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(12, 7))

    country_analysis.head(15).sort_values().plot(
        kind="barh"
    )

    plt.title(
        "Top 15 Countries by Average Renewable Energy Adoption"
    )

    plt.xlabel(
        "Average Renewable Energy Adoption"
    )

    plt.ylabel("Country")

    plt.tight_layout()
    plt.show()


# Year-wise renewable adoption
if year_col is not None:

    year_analysis = (
        df.groupby(year_col)[renewable_col]
        .mean()
        .sort_index()
    )

    plt.figure(figsize=(12, 6))

    year_analysis.plot(
        marker="o"
    )

    plt.title(
        "Global Renewable Energy Adoption Trend"
    )

    plt.xlabel("Year")
    plt.ylabel("Average Renewable Energy Adoption")

    plt.grid(True)

    plt.tight_layout()
    plt.show()


# Renewable adoption by country over time
if country_col is not None and year_col is not None:

    top_countries = (
        df.groupby(country_col)[renewable_col]
        .mean()
        .nlargest(5)
        .index
    )

    trend_df = df[
        df[country_col].isin(top_countries)
    ]

    trend = (
        trend_df.groupby(
            [year_col, country_col]
        )[renewable_col]
        .mean()
        .reset_index()
    )

    plt.figure(figsize=(13, 7))

    for country in top_countries:

        country_data = trend[
            trend[country_col] == country
        ]

        plt.plot(
            country_data[year_col],
            country_data[renewable_col],
            marker="o",
            label=country
        )

    plt.title(
        "Renewable Energy Adoption Trend - Top 5 Countries"
    )

    plt.xlabel("Year")
    plt.ylabel("Renewable Energy Adoption")

    plt.legend()

    plt.tight_layout()
    plt.show()


# Correlation Heatmap
numeric_df = df.select_dtypes(
    include=np.number
)

if numeric_df.shape[1] >= 2:

    plt.figure(
        figsize=(16, 12)
    )

    correlation = numeric_df.corr()

    sns.heatmap(
        correlation,
        cmap="coolwarm",
        center=0
    )

    plt.title(
        "Correlation Heatmap of Renewable Energy Indicators"
    )

    plt.tight_layout()
    plt.show()


print("\n" + "=" * 70)
print("CRISP-DM ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 70)
```

```python
# ============================================================
# SEMMA - GLOBAL RENEWABLE ENERGY ADOPTION ANALYSIS
# ============================================================

# ------------------------------------------------------------
# Import Libraries
# ------------------------------------------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# S - SAMPLE
# ============================================================

print("=" * 70)
print("SEMMA - S: SAMPLE")
print("=" * 70)

print("""
Sample:
Load the Global Renewable Energy and Indicators Dataset
and prepare it for analysis.
""")

uploaded = files.upload()

csv_files = [
    f for f in uploaded.keys()
    if f.lower().endswith(".csv")
]

if not csv_files:
    raise FileNotFoundError(
        "Please upload the Kaggle CSV dataset."
    )

file_name = csv_files[0]

df = pd.read_csv(file_name)

print("\nDataset loaded:", file_name)

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())

# ============================================================
# Automatically Detect Columns
# ============================================================

def find_column(columns, keywords):

    for col in columns:

        text = (
            col.lower()
            .replace("_", " ")
            .replace("-", " ")
        )

        for keyword in keywords:

            if keyword in text:
                return col

    return None


country_col = find_column(
    df.columns,
    ["country", "countries"]
)

year_col = find_column(
    df.columns,
    ["year"]
)

renewable_col = find_column(
    df.columns,
    [
        "proportion of energy from renewables",
        "renewable energy proportion",
        "renewable proportion",
        "renewable energy share",
        "renewable share",
        "renewables share",
        "renewable adoption"
    ]
)

print("\nDetected columns:")
print("Country :", country_col)
print("Year    :", year_col)
print("Target  :", renewable_col)

if renewable_col is None:

    raise ValueError(
        "Renewable-energy target column was not detected. "
        "Check the printed columns and set renewable_col manually."
    )

# ============================================================
# E - EXPLORE
# ============================================================

print("\n" + "=" * 70)
print("SEMMA - E: EXPLORE")
print("=" * 70)

print("\nData Information:")
df.info()

print("\nMissing Values:")
display(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe(include="all").T)

# ------------------------------------------------------------
# Graph 1: Renewable Adoption Distribution
# ------------------------------------------------------------

df[renewable_col] = pd.to_numeric(
    df[renewable_col],
    errors="coerce"
)

plt.figure(figsize=(10, 6))

sns.histplot(
    df[renewable_col].dropna(),
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Renewable Energy Adoption"
)

plt.xlabel(
    renewable_col
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Graph 2: Missing Values
# ------------------------------------------------------------

missing = df.isnull().sum()
missing = missing[missing > 0]

if len(missing) > 0:

    plt.figure(figsize=(12, 6))

    missing.sort_values(
        ascending=False
    ).plot(kind="bar")

    plt.title(
        "Missing Values by Column"
    )

    plt.xlabel("Column")
    plt.ylabel("Missing Values")

    plt.xticks(rotation=90)

    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# Graph 3: Global trend
# ------------------------------------------------------------

if year_col is not None:

    df[year_col] = pd.to_numeric(
        df[year_col],
        errors="coerce"
    )

    yearly = (
        df.groupby(year_col)[renewable_col]
        .mean()
        .dropna()
    )

    plt.figure(figsize=(12, 6))

    yearly.plot(
        marker="o"
    )

    plt.title(
        "Global Renewable Energy Adoption Trend"
    )

    plt.xlabel("Year")
    plt.ylabel("Average Renewable Adoption")

    plt.grid(True)

    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# Graph 4: Top Countries
# ------------------------------------------------------------

if country_col is not None:

    country_avg = (
        df.groupby(country_col)[renewable_col]
        .mean()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(12, 7))

    country_avg.head(15).sort_values().plot(
        kind="barh"
    )

    plt.title(
        "Top 15 Countries by Renewable Energy Adoption"
    )

    plt.xlabel(
        "Average Renewable Energy Adoption"
    )

    plt.ylabel("Country")

    plt.tight_layout()
    plt.show()

# ============================================================
# M - MODIFY
# ============================================================

print("\n" + "=" * 70)
print("SEMMA - M: MODIFY")
print("=" * 70)

data = df.copy()

# ------------------------------------------------------------
# Convert Target
# ------------------------------------------------------------

data[renewable_col] = pd.to_numeric(
    data[renewable_col],
    errors="coerce"
)

# Remove missing target rows
data = data.dropna(
    subset=[renewable_col]
).copy()

# ------------------------------------------------------------
# Create Binary Target
# ------------------------------------------------------------

median_value = data[renewable_col].median()

data["High_Renewable_Adoption"] = (
    data[renewable_col] > median_value
).astype(int)

print(
    "\nMedian renewable adoption:",
    median_value
)

print(
    "\nTarget counts:"
)

print(
    data["High_Renewable_Adoption"].value_counts()
)

# ------------------------------------------------------------
# Remove original target
# ------------------------------------------------------------

data.drop(
    columns=[renewable_col],
    inplace=True
)

# ------------------------------------------------------------
# Remove ID-like columns
# ------------------------------------------------------------

remove_columns = []

for col in data.columns:

    ratio = (
        data[col].nunique(dropna=False)
        / len(data)
    )

    if (
        "id" in col.lower()
        and ratio > 0.5
    ):
        remove_columns.append(col)

if remove_columns:

    print(
        "\nRemoving ID columns:",
        remove_columns
    )

    data.drop(
        columns=remove_columns,
        inplace=True
    )

# ------------------------------------------------------------
# Handle missing values
# ------------------------------------------------------------

for col in data.columns:

    if col == "High_Renewable_Adoption":
        continue

    if data[col].dtype == "object":

        mode = data[col].mode()

        if len(mode) > 0:
            data[col] = data[col].fillna(
                mode[0]
            )
        else:
            data[col] = data[col].fillna(
                "Unknown"
            )

    else:

        data[col] = data[col].fillna(
            data[col].median()
        )

# ------------------------------------------------------------
# Encode categorical columns
# ------------------------------------------------------------

encoders = {}

for col in data.select_dtypes(
    include="object"
).columns:

    encoder = LabelEncoder()

    data[col] = encoder.fit_transform(
        data[col].astype(str)
    )

    encoders[col] = encoder

# ------------------------------------------------------------
# Separate X and y
# ------------------------------------------------------------

X = data.drop(
    columns=["High_Renewable_Adoption"]
)

y = data["High_Renewable_Adoption"]

# ------------------------------------------------------------
# Convert to numeric
# ------------------------------------------------------------

X = X.apply(
    pd.to_numeric,
    errors="coerce"
)

# ------------------------------------------------------------
# Imputation
# ------------------------------------------------------------

imputer = SimpleImputer(
    strategy="median"
)

X = imputer.fit_transform(X)

# ------------------------------------------------------------
# Scaling
# ------------------------------------------------------------

scaler = StandardScaler()

X = scaler.fit_transform(X)

# ------------------------------------------------------------
# Train-Test Split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(
    "\nTraining data:",
    X_train.shape
)

print(
    "Testing data:",
    X_test.shape
)

# ============================================================
# M - MODEL
# ============================================================

print("\n" + "=" * 70)
print("SEMMA - M: MODEL")
print("=" * 70)

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

model.fit(
    X_train,
    y_train
)

print(
    "\nRandom Forest model trained successfully."
)

# ============================================================
# A - ASSESS
# ============================================================

print("\n" + "=" * 70)
print("SEMMA - A: ASSESS")
print("=" * 70)

y_pred = model.predict(
    X_test
)

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

print("\nSEMMA Model Performance")
print("-" * 40)

print(
    "Accuracy :", round(accuracy * 100, 2), "%"
)

print(
    "Precision:", round(precision * 100, 2), "%"
)

print(
    "Recall   :", round(recall * 100, 2), "%"
)

print(
    "F1 Score :", round(f1 * 100, 2), "%"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "Low Adoption",
            "High Adoption"
        ]
    )
)

# ------------------------------------------------------------
# Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Low Adoption",
        "High Adoption"
    ],
    yticklabels=[
        "Low Adoption",
        "High Adoption"
    ]
)

plt.title(
    "SEMMA - Random Forest Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Feature Importance
# ------------------------------------------------------------

feature_names = data.drop(
    columns=["High_Renewable_Adoption"]
).columns

importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    "Importance",
    ascending=False
)

print("\nTop 15 Features:")

display(
    importance.head(15)
)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=importance.head(15),
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 15 Factors Influencing Renewable Energy Adoption"
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Correlation Heatmap
# ------------------------------------------------------------

numeric = df.select_dtypes(
    include=np.number
)

if numeric.shape[1] > 1:

    plt.figure(
        figsize=(16, 12)
    )

    sns.heatmap(
        numeric.corr(),
        cmap="coolwarm",
        center=0
    )

    plt.title(
        "Correlation Heatmap - Renewable Energy Indicators"
    )

    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# Save Model
# ------------------------------------------------------------

import joblib

joblib.dump(
    model,
    "SEMMA_Renewable_Energy_RandomForest.pkl"
)

print(
    "\nModel saved successfully."
)

print("\n" + "=" * 70)
print("SEMMA ANALYSIS COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# KDD - GLOBAL RENEWABLE ENERGY ADOPTION ANALYSIS
# ============================================================

# ------------------------------------------------------------
# Import Libraries
# ------------------------------------------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# KDD STEP 1: SELECTION
# ============================================================

print("=" * 70)
print("KDD STEP 1: SELECTION")
print("=" * 70)

print("""
Select relevant data from the Global Renewable Energy
and Indicators Dataset for renewable energy adoption analysis.
""")

uploaded = files.upload()

csv_files = [
    f for f in uploaded.keys()
    if f.lower().endswith(".csv")
]

if not csv_files:

    raise FileNotFoundError(
        "Please upload the Kaggle CSV dataset."
    )

file_name = csv_files[0]

df = pd.read_csv(
    file_name
)

print(
    "\nDataset loaded:",
    file_name
)

print(
    "\nDataset shape:",
    df.shape
)

print(
    "\nFirst 5 records:"
)

display(
    df.head()
)

print(
    "\nAvailable columns:"
)

for i, col in enumerate(df.columns):

    print(
        i,
        ":",
        col
    )

# ------------------------------------------------------------
# Detect columns
# ------------------------------------------------------------

def find_column(columns, keywords):

    for col in columns:

        clean_col = (
            col.lower()
            .replace("_", " ")
            .replace("-", " ")
        )

        for keyword in keywords:

            if keyword in clean_col:

                return col

    return None


country_col = find_column(
    df.columns,
    ["country", "countries"]
)

year_col = find_column(
    df.columns,
    ["year"]
)

renewable_col = find_column(
    df.columns,
    [
        "proportion of energy from renewables",
        "renewable energy proportion",
        "renewable proportion",
        "renewable energy share",
        "renewable share",
        "renewables share",
        "renewable adoption"
    ]
)

print("\nDetected columns:")
print("Country :", country_col)
print("Year    :", year_col)
print("Target  :", renewable_col)

if renewable_col is None:

    raise ValueError(
        "Could not identify the renewable adoption column. "
        "Check the column names printed above."
    )

# ============================================================
# KDD STEP 2: PREPROCESSING
# ============================================================

print("\n" + "=" * 70)
print("KDD STEP 2: PREPROCESSING")
print("=" * 70)

data = df.copy()

# ------------------------------------------------------------
# Convert target to numeric
# ------------------------------------------------------------

data[renewable_col] = pd.to_numeric(
    data[renewable_col],
    errors="coerce"
)

# Remove records without target
data = data.dropna(
    subset=[renewable_col]
).copy()

# ------------------------------------------------------------
# Missing value analysis
# ------------------------------------------------------------

print("\nMissing values before preprocessing:")

display(
    data.isnull().sum()
)

# ------------------------------------------------------------
# Missing Value Visualization
# ------------------------------------------------------------

missing = data.isnull().sum()
missing = missing[missing > 0]

if len(missing) > 0:

    plt.figure(figsize=(12, 6))

    missing.sort_values(
        ascending=False
    ).plot(kind="bar")

    plt.title(
        "Missing Values Before Preprocessing"
    )

    plt.xlabel("Columns")
    plt.ylabel("Missing Values")

    plt.xticks(rotation=90)

    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# Create Target
# ------------------------------------------------------------

median_renewable = data[
    renewable_col
].median()

data["High_Renewable_Adoption"] = (
    data[renewable_col]
    > median_renewable
).astype(int)

print(
    "\nMedian renewable adoption:",
    median_renewable
)

# ------------------------------------------------------------
# Remove original target
# ------------------------------------------------------------

data.drop(
    columns=[renewable_col],
    inplace=True
)

# ------------------------------------------------------------
# Remove ID-like columns
# ------------------------------------------------------------

id_columns = []

for col in data.columns:

    ratio = (
        data[col].nunique(dropna=False)
        / len(data)
    )

    if (
        "id" in col.lower()
        and ratio > 0.5
    ):

        id_columns.append(col)

if id_columns:

    print(
        "\nRemoving ID columns:",
        id_columns
    )

    data.drop(
        columns=id_columns,
        inplace=True
    )

# ------------------------------------------------------------
# Handle Date columns
# ------------------------------------------------------------

for col in list(data.columns):

    if "date" in col.lower():

        date = pd.to_datetime(
            data[col],
            errors="coerce"
        )

        data[col + "_Year"] = (
            date.dt.year
        )

        data[col + "_Month"] = (
            date.dt.month
        )

        data.drop(
            columns=[col],
            inplace=True
        )

# ------------------------------------------------------------
# Fill missing values
# ------------------------------------------------------------

for col in data.columns:

    if col == "High_Renewable_Adoption":
        continue

    if data[col].dtype == "object":

        mode = data[col].mode()

        if len(mode) > 0:

            data[col] = data[col].fillna(
                mode[0]
            )

        else:

            data[col] = data[col].fillna(
                "Unknown"
            )

    else:

        data[col] = data[col].fillna(
            data[col].median()
        )

print(
    "\nMissing values after preprocessing:"
)

print(
    data.isnull().sum().sum()
)

# ============================================================
# KDD STEP 3: TRANSFORMATION
# ============================================================

print("\n" + "=" * 70)
print("KDD STEP 3: TRANSFORMATION")
print("=" * 70)

# ------------------------------------------------------------
# Encode categorical variables
# ------------------------------------------------------------

encoders = {}

categorical_columns = data.select_dtypes(
    include="object"
).columns

print(
    "\nCategorical columns:"
)

print(
    list(categorical_columns)
)

for col in categorical_columns:

    encoder = LabelEncoder()

    data[col] = encoder.fit_transform(
        data[col].astype(str)
    )

    encoders[col] = encoder

# ------------------------------------------------------------
# Separate features and target
# ------------------------------------------------------------

X = data.drop(
    columns=["High_Renewable_Adoption"]
)

y = data[
    "High_Renewable_Adoption"
]

# ------------------------------------------------------------
# Convert all features to numeric
# ------------------------------------------------------------

X = X.apply(
    pd.to_numeric,
    errors="coerce"
)

# ------------------------------------------------------------
# Remove constant columns
# ------------------------------------------------------------

constant_columns = [
    col for col in X.columns
    if X[col].nunique() <= 1
]

if constant_columns:

    print(
        "\nConstant columns removed:",
        constant_columns
    )

    X.drop(
        columns=constant_columns,
        inplace=True
    )

# ------------------------------------------------------------
# Imputation
# ------------------------------------------------------------

imputer = SimpleImputer(
    strategy="median"
)

X = imputer.fit_transform(
    X
)

# ------------------------------------------------------------
# Feature Scaling
# ------------------------------------------------------------

scaler = StandardScaler()

X = scaler.fit_transform(
    X
)

# ------------------------------------------------------------
# Train-Test Split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(
    "\nTransformed feature shape:",
    X.shape
)

print(
    "Training records:",
    X_train.shape[0]
)

print(
    "Testing records:",
    X_test.shape[0]
)

# ============================================================
# KDD STEP 4: DATA MINING
# ============================================================

print("\n" + "=" * 70)
print("KDD STEP 4: DATA MINING")
print("=" * 70)

print("""
Data Mining Method:
Random Forest Classification

Target:
High_Renewable_Adoption

0 = Low Renewable Adoption
1 = High Renewable Adoption
""")

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

model.fit(
    X_train,
    y_train
)

print(
    "\nRandom Forest training completed."
)

# ------------------------------------------------------------
# Prediction
# ------------------------------------------------------------

y_pred = model.predict(
    X_test
)

# ============================================================
# KDD STEP 5: INTERPRETATION / EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("KDD STEP 5: INTERPRETATION / EVALUATION")
print("=" * 70)

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

print("\nKDD Random Forest Results")
print("-" * 40)

print(
    "Accuracy :",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Precision:",
    round(precision * 100, 2),
    "%"
)

print(
    "Recall   :",
    round(recall * 100, 2),
    "%"
)

print(
    "F1 Score :",
    round(f1 * 100, 2),
    "%"
)

# ------------------------------------------------------------
# Classification Report
# ------------------------------------------------------------

print(
    "\nClassification Report:"
)

print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "Low Adoption",
            "High Adoption"
        ]
    )
)

# ------------------------------------------------------------
# Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(
    figsize=(7, 5)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Low Adoption",
        "High Adoption"
    ],
    yticklabels=[
        "Low Adoption",
        "High Adoption"
    ]
)

plt.title(
    "KDD - Random Forest Confusion Matrix"
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Feature Importance
# ------------------------------------------------------------

feature_names = [
    col
    for col in data.drop(
        columns=["High_Renewable_Adoption"]
    ).columns
    if col not in constant_columns
]

feature_importance = pd.DataFrame({

    "Feature": feature_names,

    "Importance":
        model.feature_importances_

})

feature_importance = (
    feature_importance
    .sort_values(
        "Importance",
        ascending=False
    )
)

print(
    "\nTop 15 Important Features:"
)

display(
    feature_importance.head(15)
)

plt.figure(
    figsize=(10, 7)
)

sns.barplot(
    data=feature_importance.head(15),
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 15 Factors Affecting Renewable Energy Adoption"
)

plt.xlabel(
    "Random Forest Feature Importance"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()
plt.show()

# ============================================================
# ADDITIONAL KDD VISUAL ANALYSIS
# ============================================================

# ------------------------------------------------------------
# Graph 1: Adoption Class Distribution
# ------------------------------------------------------------

plt.figure(
    figsize=(7, 5)
)

sns.countplot(
    x=data["High_Renewable_Adoption"]
)

plt.title(
    "High vs Low Renewable Energy Adoption"
)

plt.xlabel(
    "Adoption Class (0 = Low, 1 = High)"
)

plt.ylabel(
    "Number of Records"
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Graph 2: Global Renewable Trend
# ------------------------------------------------------------

if year_col is not None:

    trend_data = df.copy()

    trend_data[year_col] = pd.to_numeric(
        trend_data[year_col],
        errors="coerce"
    )

    trend_data[renewable_col] = pd.to_numeric(
        trend_data[renewable_col],
        errors="coerce"
    )

    trend = (
        trend_data
        .groupby(year_col)[renewable_col]
        .mean()
        .dropna()
    )

    plt.figure(
        figsize=(12, 6)
    )

    trend.plot(
        marker="o"
    )

    plt.title(
        "Global Renewable Energy Adoption Over Time"
    )

    plt.xlabel(
        "Year"
    )

    plt.ylabel(
        "Average Renewable Energy Adoption"
    )

    plt.grid(True)

    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# Graph 3: Top 15 Countries
# ------------------------------------------------------------

if country_col is not None:

    country_data = (
        df.groupby(country_col)[renewable_col]
        .mean()
        .sort_values(ascending=False)
    )

    plt.figure(
        figsize=(12, 7)
    )

    country_data.head(15).sort_values().plot(
        kind="barh"
    )

    plt.title(
        "Top 15 Countries by Renewable Energy Adoption"
    )

    plt.xlabel(
        "Average Renewable Energy Adoption"
    )

    plt.ylabel(
        "Country"
    )

    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# Graph 4: Correlation Heatmap
# ------------------------------------------------------------

numeric_data = df.select_dtypes(
    include=np.number
)

if numeric_data.shape[1] > 1:

    plt.figure(
        figsize=(16, 12)
    )

    sns.heatmap(
        numeric_data.corr(),
        cmap="coolwarm",
        center=0
    )

    plt.title(
        "Correlation Heatmap - Renewable Energy Indicators"
    )

    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# Save Model
# ------------------------------------------------------------

import joblib

joblib.dump(
    model,
    "KDD_Global_Renewable_Energy_RandomForest.pkl"
)

joblib.dump(
    scaler,
    "KDD_Renewable_Energy_Scaler.pkl"
)

joblib.dump(
    imputer,
    "KDD_Renewable_Energy_Imputer.pkl"
)

print(
    "\nModel and preprocessing objects saved successfully."
)

print("\n" + "=" * 70)
print("KDD ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 70)
```
