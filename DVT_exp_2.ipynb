```python
# Creating SQLite database

import pandas as pd
import sqlite3
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

conn = sqlite3.connect("Global_Renewable_Energy.db")

df.to_sql(
    "Renewable_Energy_Data",
    conn,
    if_exists="replace",
    index=False
)

conn.close()

print("SQLite database created successfully!")
```

```python
# Importing data from Kaggle

import pandas as pd

url = "https://www.kaggle.com/api/v1/datasets/download/anishvijay/global-renewable-energy-and-indicators-dataset"

# Download ZIP file
import requests
import zipfile
import io

response = requests.get(url)

z = zipfile.ZipFile(io.BytesIO(response.content))

print(z.namelist())

# Extract files
z.extractall("renewable_energy_data")

# Read CSV
df = pd.read_csv(
    "renewable_energy_data/complete_renewable_energy_dataset.csv"
)

print(df.head())

print(df.info())

print(df.describe())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "complete_renewable_energy_dataset.csv"
)

# Total production by energy type

production = (
    df.groupby("Energy Type")["Production (GWh)"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

plt.bar(
    production.index,
    production.values
)

plt.title(
    "Total Renewable Energy Production by Energy Type"
)

plt.xlabel("Energy Type")

plt.ylabel("Production (GWh)")

plt.xticks(rotation=45)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "complete_renewable_energy_dataset.csv"
)

# Calculate average renewable energy proportion by year

yearly_adoption = (
    df.groupby("Year")
    ["Proportion of Energy from Renewables"]
    .mean()
)

plt.figure(figsize=(10,5))

plt.plot(
    yearly_adoption.index,
    yearly_adoption.values,
    marker="o"
)

plt.title(
    "Global Renewable Energy Adoption Trend"
)

plt.xlabel("Year")

plt.ylabel(
    "Renewable Energy Share (%)"
)

plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "complete_renewable_energy_dataset.csv"
)

yearly_production = (
    df.groupby("Year")["Production (GWh)"]
    .sum()
)

plt.figure(figsize=(10,5))

plt.plot(
    yearly_production.index,
    yearly_production.values,
    marker="o"
)

plt.title(
    "Global Renewable Energy Production Trend"
)

plt.xlabel("Year")

plt.ylabel(
    "Total Production (GWh)"
)

plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "complete_renewable_energy_dataset.csv"
)

energy_trend = (
    df.groupby(
        ["Year", "Energy Type"]
    )["Production (GWh)"]
    .sum()
    .unstack()
)

plt.figure(figsize=(12,6))

for energy_type in energy_trend.columns:

    plt.plot(
        energy_trend.index,
        energy_trend[energy_type],
        marker="o",
        label=energy_type
    )

plt.title(
    "Renewable Energy Production by Source Over Time"
)

plt.xlabel("Year")

plt.ylabel(
    "Production (GWh)"
)

plt.legend()

plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "complete_renewable_energy_dataset.csv"
)

country_adoption = (
    df.groupby("Country")
    ["Proportion of Energy from Renewables"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

plt.bar(
    country_adoption.index,
    country_adoption.values
)

plt.title(
    "Top 10 Countries by Renewable Energy Adoption"
)

plt.xlabel("Country")

plt.ylabel(
    "Average Renewable Energy Share (%)"
)

plt.xticks(rotation=45)

plt.show()
```
